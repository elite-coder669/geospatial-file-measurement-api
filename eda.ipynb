```python
from shapely.geometry import LineString
import geopandas as gpd

line = LineString([
    (77.5, 12.9),
    (77.6, 13.0),
])

gdf = gpd.GeoDataFrame(
    geometry=[line],
    crs="EPSG:4326",
)

projected = gdf.to_crs(gdf.estimate_utm_crs())

print("Length in metres:", projected.geometry.length.iloc[0])
```
